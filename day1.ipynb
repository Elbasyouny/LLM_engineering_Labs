```python
# A function to display this nicely in the output, using markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

```python
# Load environment variables in a file called .env

system_promptt = "You are a professional translator working for an arabic news company to translate and summarize news web pages to arabic ignoring text that might be navigation related. Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown."
user_prompt_prefixx = """
    Translate the news to arabic
    summarize the new to arabic language
"""

# Step 2: Make the messages list

def messages_for(website):
    return [
        {"role": "system", "content": system_promptt},
        {"role": "user", "content": user_prompt_prefixx + website}
    ]

# Step 3: Call OpenAI
def summarize(url):
    website = fetch_website_contents(url)
    response = openai.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = messages_for(website)
    )
    return response.choices[0].message.content

# Step 4: print the result
display_summary("https://BBC.com")
```
