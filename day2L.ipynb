```python
OLLAMA_BASE_URL = "http://localhost:11434/v1"

ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')
```

```python
def summarize(url):
    website = fetch_website_contents(url)
    response = ollama.chat.completions.create(
        model = "llama3.2",
        messages = messages_for(website)
    )

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))

system_promptt = "You are a professional translator working for an arabic news company to translate and summarize news web pages to arabic including important news and ignoring text that might be navigation related. Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown."
user_prompt_prefixx = """
    Translate the news to arabic
    summarize the news to prefecint arabic language including all important details
"""

# Step 2: Make the messages list

def messages_for(website):
    return [
        {"role": "system", "content": system_promptt},
        {"role": "user", "content": user_prompt_prefixx + website}
    ]

# Step 3: Call OpenAI
def summarize(url):
    website = fetch_website_contents(url)
    response = ollama.chat.completions.create(
        model="llama3.2",
        messages = messages_for(website)
    )
    return response.choices[0].message.content

# Step 4: print the result
display_summary("https://cnn.com")
```
